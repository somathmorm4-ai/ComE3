## Exercise 1: Basic try/except

1. Write code that divides `10` by `0` inside a `try` block.
2. Catch the error in `except` and print a friendly message.
3. After the block, print `"Program continues"`.

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")

print("Program continues")
```

## Exercise 2: else and finally

1. In a `try` block, divide `20 / 4` and print result.
2. Add an `except` block for errors.
3. Add an `else` block that prints `"No errors"`.
4. Add a `finally` block that prints `"Finished"`.

```python
try:
    result = 20 / 4
    print("Result:", result)
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("No errors")
finally:
    print("Finished")
```

## Exercise 3: None

1. Create `value = None`.
2. Check with `is None` and print `"No value"` when true.
3. Set `value = "Python"`.
4. Check again and print `"Has value"` when it is not None.

```python
value = None
if value is None:
    print("No value")

value = "Python"
if value is not None:
    print("Has value")
```

## Exercise 4: Input and casting

1. Ask user input for name.
2. Ask user input for age.
3. Convert age to integer.
4. Print: `"Hello <name>, next year you will be <age+1>"`.
5. Wrap conversion in `try/except` to catch invalid age input.

```python
name = input("Enter your name: ")
age_text = input("Enter your age: ")

try:
    age = int(age_text)
    print(f"Hello {name}, next year you will be {age + 1}")
except ValueError:
    print("Invalid age: please enter a whole number")
```

## Exercise 5: Safe mini calculator

1. Ask two numbers from user input.
2. Convert both to `float`.
3. Print sum, subtraction, multiplication, and division.
4. Handle invalid numbers with `ValueError`.
5. Handle divide-by-zero with `ZeroDivisionError`.

```python
try:
    first = float(input("Enter the first number: "))
    second = float(input("Enter the second number: "))
    print("Sum:", first + second)
    print("Subtraction:", first - second)
    print("Multiplication:", first * second)
    print("Division:", first / second)
except ValueError:
    print("Invalid number input")
except ZeroDivisionError:
    print("Cannot divide by zero")
```
