## Exercise 1: Variable swap and trace

- Create `a = 15` and `b = 40`.
- Swap them **without** using a third variable.
- Print before/after values in a clear format.
- Print the type of each variable after swapping.

```python
a = 15
b = 40
print("Before swap: a =", a, "b =", b)

a, b = b, a
print("After swap: a =", a, "b =", b)
print("Type of a:", type(a))
print("Type of b:", type(b))
```

## Exercise 2: Dynamic typing checkpoint

- Create one variable `v` and assign it sequentially: string, int, float, bool, list.
- After each assignment, print the value and its type.
- Explain in a comment what this shows about Python variables.

```python
v = "hello"
print(v, type(v))

v = 42
print(v, type(v))

v = 3.14
print(v, type(v))

v = True
print(v, type(v))

v = [1, 2, 3]
print(v, type(v))

# Python variables are dynamically typed: the same name can point to
# objects of different types, and the type belongs to the value, not the name.
```

## Exercise 5: Type-driven expression building

- Create variables for name (str), age (int), height (float), active (bool).

- Print the type of every variable at the end.
