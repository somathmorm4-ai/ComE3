## Exercise 1: Basic functions

1. Define a function `greet()` that prints `"Hello from function"`.
2. Call `greet()` two times.
3. Define `square(n)` that returns `n * n`.
4. Call `square(5)` and print the result.

```python
def greet():
    print("Hello from function")

greet()
greet()

def square(n):
    return n * n

print(square(5))
```

## Exercise 2: Function arguments

1. Define `introduce(name, age)` and print: `"My name is <name> and I am <age>"`.
2. Call it once with positional arguments.
3. Call it once with keyword arguments.
4. Add a default value for age and call the function with only name.

```python
def introduce(name, age=18):
    print(f"My name is {name} and I am {age}")

introduce("Anna", 21)
introduce(name="Ben", age=22)
introduce("Chenda")
```

## Exercise 3: *args

1. Define `total(*numbers)` that returns the sum of all numbers.
2. Call it with 3 numbers and print the result.
3. Call it with 5 numbers and print the result.

```python
def total(*numbers):
    return sum(numbers)

print(total(1, 2, 3))
print(total(1, 2, 3, 4, 5))
```

## Exercise 4: Scope

1. Create `msg = "Global"` outside any function.
2. Define a function that creates `msg = "Local"` and prints it.
3. Print `msg` outside the function again.
4. Observe and explain (in a comment) why output is different.

```python
msg = "Global"

def show_msg():
    msg = "Local"
    print(msg)

show_msg()
print(msg)

# The msg inside show_msg belongs to the function's local scope, so it is a
# different variable from the global msg and does not change it.
```

## Exercise 5: Lambda and recursion

1. Create a lambda function `double` that returns `x * 2`.
2. Print `double(9)`.
3. Write a recursive function `factorial(n)`.
4. Print `factorial(5)`.

```python
double = lambda x: x * 2
print(double(9))

def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```
