## Exercise 1: Truth table mini-lab

- Create booleans `p = True`, `q = False`.
- Print results for: `p and q`, `p or q`, `not p`, `not q`.
- Add one compound expression with parentheses and print its result.

```python
p = True
q = False

print("p and q:", p and q)
print("p or q:", p or q)
print("not p:", not p)
print("not q:", not q)

print("(p or q) and not q:", (p or q) and not q)
```

## Exercise 3: Operator precedence check

- Create a complex boolean expression using `and`, `or`, and `not`.
- Evaluate it once without extra parentheses and once with explicit parentheses.
- Print both and comment whether they match.

```python
a = True
b = False
c = True

without_parens = a or b and not c
with_parens = a or (b and (not c))

print("Without extra parentheses:", without_parens)
print("With explicit parentheses:", with_parens)

# They match because `not` binds before `and`, and `and` binds before `or`,
# so the explicit parentheses only make that default order visible.
```

## Exercise 4: Identity vs equality

- Create two equal lists and compare with `==` and `is`.
- Assign one list to another variable and compare again.
- Write a short comment explaining differences in output.

```python
list_a = [1, 2, 3]
list_b = [1, 2, 3]

print("list_a == list_b:", list_a == list_b)
print("list_a is list_b:", list_a is list_b)

list_c = list_a
print("list_a == list_c:", list_a == list_c)
print("list_a is list_c:", list_a is list_c)

# `==` compares values, so equal contents give True.
# `is` compares object identity, so list_b is a different object
# while list_c points to the same object as list_a.
```

## Exercise 5: Membership filter

- Given `items = ["python", "java", "go", "rust"]`, test membership for 4 queries.
- Use both `in` and `not in`.
- Print human-readable results like `python: found`.

```python
items = ["python", "java", "go", "rust"]

for item in ["python", "javascript", "go", "c++"]:
    if item in items:
        print(f"{item}: found")
    if item not in items:
        print(f"{item}: not found")
```
